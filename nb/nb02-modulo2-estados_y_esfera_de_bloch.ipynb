## Estado a partir de los ángulos 

Una forma de construir un estado normalizado es usando la parametrización 

$$
|u\rangle = \cos{\frac{\theta}{2}}\, |0\rangle + e^{i\phi}\sin{\frac{\theta}{2}}\,|1\rangle 
$$

Hagamos una función que nos de el estado a partir de los ángulos $\theta$ y $\phi$

```python
def state_from_angles(theta, phi):
    u0 = np.cos(theta/2)
    u1 = (np.exp(1j*phi))*np.sin(theta/2)

    uket = [u0,u1]
    return Statevector(uket)
statevector_y_plus = state_from_angles(np.pi/2, np.pi/2)
statevector_y_plus.draw('Latex')
```

## Producto tensorial (de Kronecker)

```python
# Definimos una función para visualizar el estado y la esfera a la vez
# Permite escalar la(s) esferas(s) con 'inches' pulgadas, si este parámetro
# no se pone, se dibuja al tamaño por defecto
# Si se pone 0, no se dibuja la esfera
def my_display_state_vector(psi, inches=None):
    d_latex = psi.draw('latex', prefix=r"|\psi\rangle = ")
    display(d_latex)
    #
    d_bloch = psi.draw('bloch')
    if isinstance(inches, (int,float)):
        if inches > 0:
            d_bloch.set_size_inches(inches, inches, forward=True)
            display(d_bloch)
    else:
        display(d_bloch)

# Generamos dos vectores de estado
v1 = np.array([-1, 0.5])
v2 = np.array([1, 1-1j])
s1 = Statevector(v1/la.norm(v1))
s2 = Statevector(v2/la.norm(v2))
                
my_display_state_vector(s1, 2)
my_display_state_vector(s2, 2)
```

```python
# Vamos a ver diferentes formas de realizar el producto de kronecker

# La primera, usando el operador circunflejo ^
sc1 = s1 ^ s2
sc2 = s1 ^ s2 ^ s2

# Dibujar las multiesferas
my_display_state_vector(sc1,3)
my_display_state_vector(sc2,4.25)
```

```python
# Usando el método tensor()
#  self.tensor(other) = kron(self, other)

sc1_t = s1.tensor(s2)
sc2_t = s1.tensor(s2.tensor(s2))

my_display_state_vector(sc1_t, 0)
my_display_state_vector(sc2_t, 0)
```

```python
# Usando el método expand()
#   self.expand(other) = kron(other,self)

sc1_e = s2.expand(s1)
sc2_e = (s2.expand(s2)).expand(s1)

my_display_state_vector(sc1_e, 0)
my_display_state_vector(sc2_e, 0)
```

```python
# Comprobemos que todas las formas de calcular el producto
# tensorial dan el mismo resultado

assert(sc1 == sc1_e)
assert(sc2 == sc2_e)
assert(sc1 == sc1_t)
assert(sc2 == sc2_t)
```

## Representación estados no factorizables

Veamos cómo se pueden representar los estados no factorizables _(entangled)_ visualmente.

```python
e0 = Statevector([1, 0], dims=2)  # dims indica la dimensión del estado
e1 = Statevector([0, 1], dims=2)

ebell = ((e0^e0)+(e1^e1))/np.sqrt(2)
# Alternativa
#ebell = Statevector([1/sqrt(2), 0, 0, 1/sqrt(2)])

# Como hay entanglement, ¿se pueden representar los qubits por separado?
# Observa el resultado de la(s) esfera(s) de Bloch
my_display_state_vector(ebell,3)
```

```python
# En estos casos, una forma de visualizar el estado (en Qiskit)
# es usando una Q-esfera.
#
# Cada amplitud se representa por un circulo, cuyo diámetro es proporcional a
# la magnitud de la amplitud. La fase se representa mediante un código de colores.
display(plot_state_qsphere(ebell, figsize=(4,4)))
```

```python
# Mostremos la Q-sphere de un estado con aspecto extraño
emix = e0 + (2+2j)*e1
eleftminus = Statevector.from_label('l') + Statevector.from_label('-')
strangestate = emix^ebell^eleftminus 
strangestate = strangestate/la.norm(strangestate)
display(strangestate.draw('latex'))
plot_state_qsphere(strangestate, figsize=(4,4))
```

## Revisitando la evolución de un estado en la esfera de Bloch


### Evolución temporal de un sistema de un qubit (dimensión 2), representado en la esfera de Bloch

<!-- div class="alert alert-block alert-success" -->
<div>
La evolución de la ecuación de estado a lo largo del tiempo dado un hamiltoniano $H$ independiente del tiempo viene dada por

$$
\boxed{\psi(t)= e^{{\Large \frac{-i}{\hbar}} t H} |\psi_0\rangle \text{, donde para t=0:} \,\, \psi(0) = |\psi_0\rangle}
$$

Escribe una función <i>psi(t,psi_0_ket,H)</i> que reciba como argumentos el array <i>psi0</i> de componentes de $|{\psi(0)}\rangle$, una matriz hermítica <i>H</i>, y un valor de <i>t</i>, y devuelva un vector columna correspondiente a $|{\psi(t)}\rangle$. Considera $\hbar=1$.

</div>

```python
# Nuestra constante de Plank "h barra = h/(2*pi)"
hbar = 1

def expH(t, H):
    # Retornamos la exponencial de la ecuación de ondas
    #    |\psi(t)> = e^{-i t H/hbar} |\psi(0)> = U |\psi(0) >
    # o sea,
    #    U = e^{-i t H/hbar}
    #
    U = la.expm(-1j*t*H/hbar);
    return U

def psi(t, psi_0_ket, H):
    # Retornamos la función de ondas
    #    |\psi(t)> = e^{-i t H/hbar} |\psi(0)> = U |\psi(0) >
    #
    lpsi = len(psi_0_ket)
    lH = H.shape[0]
    assert lpsi == lH # Chequear dimensión

    # |\psi(t)> = U |\psi(0) >
    # U = e^(- 1j*t*H/hbar);
    U = expH(t, H)
    psi_t_ket = U @ psi_0_ket
   
    return psi_t_ket
```

Ahora:
- Partiendo del instante 0 y para un intervalo de tiempo razonable, muestrea el valor del estado en cada instante
- Crea una animación con la representación del vector de estado en la esfera de Bloch en cada instante

Se considerará que:
- El Hamiltoniano es una combinación lineal de matrices de Pauli 2x2; se puede empezar probando con la matriz Z
- El valor inicial del estado del sistema, para $t=0$, debe ser un estado que NO sea autovector de la matriz elegida

```python
# Vectores |0> y |1> de la base computacional por si hicieran falta
zero_ket = np.array([[1], [0]])
one_ket = np.array([[0], [1]])

# Combinación lineal real de matrices de Pauli
def H_pauli(a1=0, a2=0, a3=1):
    X=np.array([[0,1], [1,0]])
    Y=np.array([[0,-1j], [+1j,0]])
    Z=np.array([[1,0], [0,-1]])
    return a1*X + a2*Y + a3*Z

H = H_pauli(0,0,1)
print('H=', H)

# Fijamos estado inicial que no sea autovector de H
psi0 = zero_ket - 1/3j*one_ket
psi0 = psi0/la.norm(psi0)
```

```python
# Vamos a estimar la frecuencia asociado al Hamiltoniano para
# que luego las gráficas tengan un mejor aspecto
#
# Estimación de la frecuencia asociada al hamiltoniano, como
# e^(jwt)=cos(wt)+jsin(wt) usamos el máximo de los elementos
freq = np.max(np.abs(H))/2/np.pi/hbar
T=1/freq
print("Estimación de frecuencia={0:.2f}, w=2*pi*f={1:.2f}, T={2:.2f}".format(freq, 2*np.pi*freq, T))
```

```python
# Elegimos aproximadamente N periodos, con samples muestras por periodo
N = 2; samples = 17
t = np.linspace(0, T*N, samples*N)  # time axis
print("#samples="+str(len(t)))
```

```python
# Valores de |Psi(t)>
psi_t_ket = [psi(xt, psi0, H) for xt in t] # |psi(t)>

# Creamos una secuencia de ficheros PNG que guardamos
c=0
l=[]
workdir='/tmp'
for p in psi_t_ket:
    psi_s = Statevector(p)
    bloch = psi_s.draw('bloch', figsize=(2.5,2.5))
    filename = workdir + '/' + 'psi_bloch_%04d.png'%(c)
    bloch.savefig(filename)
    c += 1
    l.append(filename)
    #display(bloch)

# Check files
#!ls -l {workdir}/*png
#display(l)
```

```python
# Crear animación

import matplotlib.pyplot as plt
import matplotlib.animation as animation
from matplotlib import image

def my_animate_files(imgfilelist):
    fig, ax = plt.subplots()

    # Function to update the image in the animation
    img = ax.imshow(image.imread(imgfilelist[0]))
    def update_img(i):
        img.set_data(image.imread(imgfilelist[i]))
        return img
    
    ani = animation.FuncAnimation(fig, update_img, frames=len(imgfilelist), interval=100, repeat=False)
    #plt.show()
    #plt.axis('off')
    plt.close() # Avoid showing the static figure at the end
    
    return HTML(ani.to_jshtml())

html = my_animate_files(l)
display(html)
```
