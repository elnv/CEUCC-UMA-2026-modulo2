````python
# Utilería acumulada

# Definimos una función para visualizar el estado y la esfera a la vez
# Permite escalar la(s) esferas(s) con 'inches' pulgadas, si este parámetro
# no se pone, se dibuja al tamaño por defecto
# Si se pone 0, no se dibuja la esfera
def my_display_state_vector(psi, inches=None, prefix=r"|\psi\rangle = "):
    if isinstance(psi, Statevector):
        d_latex = psi.draw('latex', prefix=prefix)
        display(d_latex)
        #
        d_bloch = psi.draw('bloch')
        if isinstance(inches, (int,float)):
            if inches > 0:
                d_bloch.set_size_inches(inches, inches, forward=True)
                display(d_bloch)
        else:
            display(d_bloch)

#----------------------------------------------------------

# Función para simular un circuito con el método "statevector"
# Devuelve el resultado de la simulación
def my_circuit_simulate_statevector(qcirc, shots=1):
    # Instancia del simulador
    sim = AerSimulator(method = "statevector")
     
    # Hacemos una simulación y obtenemos los resultados
    qobj = transpile(qcirc, backend=sim)
    result = sim.run(qobj, shots=shots).result()
    
    return result
    
# Función para simular un circuito con el método "unitary"
# Devuelve el resultado de la simulación
def my_circuit_simulate_unitary(qcirc, shots=1):
    # Instancia del simulador
    sim = AerSimulator(method = "unitary")
     
    # Hacemos una simulación y obtenemos los resultados
    qobj = transpile(qcirc, backend=sim)
    result = sim.run(qobj, shots=shots).result()
    
    return result
    
#----------------------------------------------------------

# Función para obtener el vector de estado final de un cirucito dado
# usando la simulación de vector de estado.
# Devuelve esta tupla de 3 elementos:
#         (estado final, cuenta, resultado)
def my_get_circuit_state(qcirc_in, shots=1, final_probe=True):
    # Hacemos una copia ya que vamos a cambiar el objeto circuito
    qcirc = qcirc_in.copy()
    
    # Ponemos el "probe" para el vector de estado al final
    # del cirucito dado
    # ojo! el circuito no debe de tener un "save_statevector" ya!
    if final_probe:
        qcirc.save_statevector()
    
    # Hacemos una simulación y obtenemos los resultados
    result = my_circuit_simulate_statevector(qcirc, shots)
    state = None
    if final_probe:
        state = result.get_statevector()
    counts = result.get_counts()
    
    return state, counts, result

# Función para obtener la matriz unitaria final de un cirucito dado
# usando la simulación de matriz unitaria.
# Devuelve esta tupla de 2 elementos:
#         (unitary, resultado)
def my_get_circuit_unitary(qcirc_in, shots=1, final_probe=True):
    # Hacemos una copia ya que vamos a cambiar el objeto circuito
    qcirc = qcirc_in.copy()
    
    # Ponemos el "probe" para el vector de estado al final
    # del cirucito dado
    # ojo! el circuito no debe de tener un "save_statevector" ya!
    if final_probe:
        qcirc.save_unitary()
    
    # Hacemos una simulación y obtenemos los resultados
    result = my_circuit_simulate_unitary(qcirc, shots)
    unitary = result.get_unitary()
    
    return unitary, result
````

## Circuito de teleportación

````python
from qiskit.quantum_info import partial_trace

# Preparación del estado a teleportar 
# Vamos a "preparar" un estado |φ⟩ = Ry(π/4)|0⟩ mediante un circuito 
state_prep_circ = QuantumCircuit(1, name="φ_prep")
initial_angle = np.pi / 4
state_prep_circ.ry(initial_angle, 0)

# Este es el vector de estado perparado, esta variable es solo
# para visualizar y comparar
initial_state = Statevector.from_instruction(state_prep_circ)

# Alternativa: Inicialización arbitraria, usar el método
# initialize() en lugar de añadir el mini-circuito anterior preparador
#state0 = np.array([0.7, 0.3-1j])
#state0 = state0/la.norm(state0)
#initial_state = Statevector(state0)
    
print("Estado inicial |φ⟩ listo para teleportarse:")
my_display_state_vector(initial_state,2.5)

# Vamos a construir el circuito de teletransporte: 
#  - 3 qubits + 2 bits clásicos para medir
# q0: Bit a teleportar: |φ⟩ (Alice)
# q1: Mitad del estado de Bell de Alice
# q2: Mitad del estado de Bell de Bob
# c0: Bit clasico 'x' para medir q0 
# c1: Bit clásico 'y' para medir q1 

q = QuantumRegister(3, name="q")
c0 = ClassicalRegister(1, name="x")
c1 = ClassicalRegister(1, name="y")
teleport_circ = QuantumCircuit(q, c0, c1)

# A. Inicializar q0 con |φ⟩ (Alice) 
teleport_circ.append(state_prep_circ, [q[0]]) # Inicialización mediante circuito
#teleport_circ.initialize(initial_state, [q[0]]) # Alternativa: inicializar con estado arbitrario, no con circuito
teleport_circ.barrier()

# B. Crear el estado de Bell |B₀₀⟩ compartido entre Alice (q1) y Bob (q2)
teleport_circ.h(q[1])
teleport_circ.cx(q[1], q[2])
teleport_circ.barrier()

# C. Desentrelazado en el lado de Alice (CNOT + H)
teleport_circ.cx(q[0], q[1])
teleport_circ.h(q[0])
teleport_circ.barrier()

# D. Alice mide sus qubits 
teleport_circ.measure(q[0], c0)
teleport_circ.measure(q[1], c1)
teleport_circ.barrier()

# E. Bob aplica las puertas cuánticas clásicamente controladas
# Esto son las operaciones Z^x y X^y
teleport_circ.z(q[2]).c_if(c0, 1) # Aplicar Z si x==1
teleport_circ.x(q[2]).c_if(c1, 1) # Aplicar Y si y==1

# Versión alternativa de las puertas controladas clásicamente:
# Uso de los "ifs dinámicos"
# (solo funciona para backend "modernos")
#with teleport_circ.if_test((c0, 1)):
#    teleport_circ.z(q[2])
#
#with teleport_circ.if_test((c1, 1)):
#    teleport_circ.x(q[2])

# Circuito de teletransporte 
print("Circuito de teletransportación")
display(teleport_circ.draw('mpl', style='iqp', scale=0.7))

# Simulación 
s, c, rs = my_get_circuit_state(teleport_circ, shots=1000)
final_statevector = s
print("Estado final del circuito")
my_display_state_vector(final_statevector,7)

# Para tener el estado final del qubit teletransportado 
# vamo a aplicar una operación que deshace el producto tensorial 
# Esto se llama TRAZA PARCIAL
# Hay que tener en cuenta que los qubits están ordenados así |q2 q1 q0⟩
# El método de traza parcial retorna un objeto MATRIZ DE DENSIDAD
bobs_final_density_mat = partial_trace(final_statevector, [0, 1]) 

# Convertir matriz de densidad a statevector
bobs_final_state = bobs_final_density_mat.to_statevector()

# Este ya es el estado teletransportado en posesión de Bob
print("\nBob's final state on his qubit (q2):")
my_display_state_vector(bobs_final_state,3)
# display(plot_bloch_multivector(bobs_final_densitymat)) # alternativa visualizar matriz de densidad como state vector

# Chequeo de que  el estado inicial |φ⟩ y el teletransportado son el mismo
ok = bobs_final_state.equiv(initial_state)
print(f"Teletransporte OK (¿estado de Bob y original son iguales?): {ok}")
````

### Medidas de calidad de un circuito 

Alguna medidas cuantitativas  permiten comparar la calidad de distintos circuitos que efectúan la misma tarea. 

- **Anchura:** es el número total de cúbits  que necesita.  El uso de ancillas incrementa la anchura de un circuito, y por tanto, reduce su calidad en comparación con otro circuito que tenga menor anchura. 

- **Coste:** el número de puertas presentes en el mismo. 

- **Complejidad:** es una medida estandarizada asociada al número de *puertas elementales* en las que puede descomponerse un circuito.  Es un número a reducir. Sin embargo no es una medida inambigua ya que depende de la librería utilizada. Por ejemplo, si ésta es la NCT, entonces el coste del es 1. Sin embargo, si la librería es la tomada por $ \{ H, S, T, \hbox{CNOT}\} $
entonces el coste sube hasta 7. Por ello, a la hora de comparar circuitos es importante definirlos en la misma base. 

- **Profundidad:** para evaluar la *profundidad*  es necesario agrupar todas las puertas  que se puedan realizar en paralelo en cortes temporales de duración $\Delta$ (pulso). En particular puertas que actúen sobre registros diferentes no interferirán y se podrán paralelizar.

## _Phase kick-back_

Cuando la puerta es controlada de fase, el bit controlador y controlado son
intercambiables.

Es un caso particular del fenómeno conocido como Kick-Back ¿qué interpretación tiene?


````python
# PHASE KICK-BACK

# Creación de una nueva puerta: 
    # Vamos a crear una puerta controlada de fase P con
    # ángulo theta = pi/4

theta=np.pi/4  # Phase for controlled phase gate, when theta=pi, is the controlled-Z matrix 

c=0 # control bit
t=1 # target bit

# 1ª forma de crear una puerta: a partir de un circuito
#    usando la puerta P controlada de qiskit
qcgate = QuantumCircuit(2, name='MY1')
qcgate.cp(theta, c, t)  # Añadir puerta de fase controlada de qiskit
my_gate1 = qcgate.to_gate()

# 2ª forma, usar una puerta y convertirla en controlada
qcgate = QuantumCircuit(1, name='MY2')
qcgate.p(theta, 0)
my_gate2 = qcgate.to_gate()
my_gate2 = my_gate2.control(1, label="CMY")

# Este es un circuito base con dos puertas
qc=QuantumCircuit(2)
qc.h(c)
qc.x(t)

# Vamos a añadir la puerta MY1 que es una puerta 
# de dos qubits
qc1 = qc.copy()
qc1.append(my_gate1,[0,1])

# Vamos a añadir la puerta MY2 que es una puerta
# controlada (también de dos qubits)
qc2 = qc.copy()
qc2.append(my_gate2,[0,1])

# Vamos a añadir la puerta MY2 controlada
# al revés
qc3 = qc.copy()
qc3.append(my_gate2,[1,0])

# Simulemos
s1, c1, r1 = my_get_circuit_state(qc1, shots=1)
s2, c2, r2 = my_get_circuit_state(qc2, shots=1)
s3, c3, r3 = my_get_circuit_state(qc3, shots=1)

# Visualizemos los circuitos y la salida de la simulación
display(qc1.draw('mpl', style="iqp"))
my_display_state_vector(s1, 3)

display(qc2.draw('mpl', style="iqp"))
my_display_state_vector(s2, 3)

display(qc3.draw('mpl', style="iqp"))
my_display_state_vector(s3, 3)
````

## Transformación cuántica que implementa una función lógica

En este ejemplo se costruye una puerta cuántica que implementa una fución lógica $f(x)$ dada.

Vamos a considerar que tanto la entrada $x$ de la función como su salida $f(x)$ son valores de dos bits.

Una característica importante a observar es **la reversibilidad** de las puertas cuánticas, ya que son operadores unitarios

En este ejemplo se trata de construir una puerta reversible dada una función f(x):
 
$$     
    U: |x\rangle|y\rangle \rightarrow |x\rangle|y \oplus f(x)\rangle,~~~ \text{ con $\oplus$ el operador XOR=bitwise exclusive OR}
$$
     
```text
               2    +---+    2               
         |x> --/----|   |----/-- |x>         
                    | U |                  
         |y> --/----|   |----/-- |x ⊕ f(x)>
               2    +---+    2               
```      

En este ejemplo:
 
$$
f(x_1,x_0) = (x_1, x_1 \oplus x_0)
$$

Donde: 
 - $|x\rangle = |x1\rangle|x0\rangle$            de dimensión dos (2 qubits)
 - $|f(x)\rangle = |f(x)[1]\rangle |f(x)[0]\rangle$ también de dimensión 2

Esta es la tabla de verdad asociada:

```text
 x1 x0  f(x1,x0)[1]  f(x1,x0)[0]
 -----  -------------------
  0 0             0 0
  0 1             0 1
  1 0             1 1
  1 1             1 0
```
  
 Observar que se añaden 2 qubits ancilla $|y\rangle$ que se inicializan
 al estado 0, por lo que a la salida final, esos 2 qubits ancilla
 se han transformado  en $|y \oplus f(x)\rangle = |0 \oplus f(x)\rangle = |f(x)\rangle$.

 Aquí evaluaremos el ciruito con los posibles valores
 de inicialización de $|x\rangle: |00\rangle, |01\rangle, |10\rangle, |11\rangle$

````python
"""
 Ejemplo de construir una PUERTA REVERSIBLE dada una función f(x):
"""

# Definimos un registro cuántico con un qúbit
quantum_reg_x = QuantumRegister(2, 'x') 
quantum_reg_y = QuantumRegister(2, 'y') # Ancilla for reversibiity

# Vamos a crear un circuito para cada valor de |x1x0>
for i in range(0,2**2):
    # Creamos el circuito pasandole los registros. 
    qc = QuantumCircuit(quantum_reg_x, quantum_reg_y)

    # Notar que x0 es el menos significativo de la palabra
    # aunque aparezca el más arriba en el diagrama
    # Y por tanto los qubit ancilla resultan los más significativos
    # (van más a la izquierda en la palabra multiqubit)

    # Inicializar x a i
    qc.initialize(i, quantum_reg_x[0:3])

    # Inicializar |y> (ancilla) a 0
    qc.initialize(0, quantum_reg_y[0])
    qc.initialize(0, quantum_reg_y[1])

    # Puertas
    qc.cx(quantum_reg_x[1], quantum_reg_y[1])
    qc.cx(quantum_reg_x[0], quantum_reg_y[0])
    qc.cx(quantum_reg_x[1], quantum_reg_y[0])

    # Dibujamos el circuito
    display(Markdown("**Circuito para $|x\\rangle$ = $|{i}\\rangle$**".format(i=i)))
    display(qc.draw(output='mpl', style='iqp', reverse_bits=False, scale=0.75))
    
    # Simulamos
    #result = sim.run(qc, shots = 1).result()
    s, c, r = my_get_circuit_state(qc, shots=1)
    result_state = s
    
    x0= (i & 1)      # bits de i para poner en el prefijo
    x1= (i & 2)>>1
    display(result_state.draw('latex', prefix=r"|f({x1:d}{x0:d})\rangle |{x1:d}{x0:d}\rangle \Rightarrow |y1\rangle |y0\rangle |x1\rangle |x0\rangle = ".format(x0=x0,x1=x1)))
    display(Markdown('---'))
````

````python
def linear_circuit(x,a):
       
    assert(len(x)==len(a))

    # Inicialización de los registros
    qr_in = QuantumRegister(len(a), name='qr_in')
    qr_out = QuantumRegister(1, name='qr_out')
    cr = ClassicalRegister(1, name='cr')  
    qc = QuantumCircuit(qr_in, qr_out, cr, name='q_linear')
    
    'inicializamo el estado x '
    # Recorremos la cadena x y vamos aplicando puertas X donde haya un 1 para inicializar el circuito
    #for i, xq in enumerate(reversed(x)):  # ojo con la ordenación de qiskit, por eso está reversed()
    for i in range(len(x)):
        xq = x[len(x)-1 -i] # Recorremos x al reves porque la ordenación de qiskit es al reves      
        print(i, xq)
        if xq == '1':
             qc.x(qr_in[i]) 

    qc.barrier()

    'codificamos la función lineal x.a '
###
#
#        Tu solución aquí
#
#
####
            
    qc.barrier()
    qc.measure(qr_out[0],cr[0])
    
    return qc 
````

````python
a = '1011'
x = '1101'

circuit = linear_circuit(x,a)
circuit.draw('mpl', style="iqp")
````

La función $a\cdot x = (1 + 0 + 0 + 1)mod(2) = 0$. Vamos a ver si este resultado es el hallado

````python
# transpilamos
t_circuit = transpile(circuit, backend = simulador)

# Ejecutamos la simulación con 1000 shots 
result = simulador.run(t_circuit, shots = 1000).result()
counts = result.get_counts()
counts
````
